## Problem 1 LRU Cache Implementation

To implement the LRU (Least Recently Used) Cache efficiently, I used a combination of a Hash Map and Doubly Linked List.

Hash Map allows direct access to cache items using their keys in O(1) time and  Doubly Linked List keeps track of the order in which items are used. The most recently used item is placed at the front whereas the least recently used item remains at the end.

When a key is accessed through the get() operation, it is moved to the front of the list. When a new item is added and the cache is already full, the item at the end of the list is removed because it is the least recently used.

This combination ensures that both get() and put() operations run in O(1) time.

```python
class Node:
    def __init__(self, key, value):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None


class LRUCache:

    def __init__(self, capacity):
        self.capacity = capacity
        self.cache = {}

        self.head = Node(0, 0)
        self.tail = Node(0, 0)

        self.head.next = self.tail
        self.tail.prev = self.head

    def remove(self, node):
        prev_node = node.prev
        next_node = node.next

        prev_node.next = next_node
        next_node.prev = prev_node

    def insert(self, node):
        node.next = self.head.next
        node.prev = self.head

        self.head.next.prev = node
        self.head.next = node

    def get(self, key):
        if key not in self.cache:
            return -1

        node = self.cache[key]
        self.remove(node)
        self.insert(node)

        return node.value

    def put(self, key, value):

        if key in self.cache:
            self.remove(self.cache[key])

        node = Node(key, value)
        self.cache[key] = node
        self.insert(node)

        if len(self.cache) > self.capacity:
            lru = self.tail.prev
            self.remove(lru)
            del self.cache[lru.key]
```

The LRU Cache efficiently stores frequently used data and removes the least recently used item when the cache reaches its capacity.

```python
cache = LRUCache(2)

cache.put(1, 10)
cache.put(2, 20)

print(cache.get(1))
```

## Problem 2 Meeting Scheduler

#### Checking if All Meetings Can Be Attended
The meetings are first sorted according to their start time. After sorting, each meeting is compared with the previous meeting. If the start time of the current meeting is earlier than the end time of the previous meeting, an overlap exists and the person cannot attend all meetings.

If no overlap is found, the person can attend every meeting.

#### Finding Minimum Rooms Required
If a meeting starts before another meeting ends, an additional room is needed.
Otherwise, an existing room can be reused.

```python
def can_attend_all(events):
    events.sort(key=lambda x: x[0])

    for i in range(1, len(events)):
        if events[i][0] < events[i - 1][1]:
            return False

    return True


def min_rooms_required(events):
    if not events:
        return 0

    starts = sorted([event[0] for event in events])
    ends = sorted([event[1] for event in events])

    rooms = 0
    max_rooms = 0
    s = e = 0

    while s < len(events):
        if starts[s] < ends[e]:
            rooms += 1
            max_rooms = max(max_rooms, rooms)
            s += 1
        else:
            rooms -= 1
            e += 1

    return max_rooms


events = [(9, 10), (10, 11), (11, 12)]

print("Can Attend All Meetings:", can_attend_all(events))
print("Minimum Rooms Required:", min_rooms_required(events))
```

There is no overlap between the meetings because each meeting starts exactly when the previous one ends(a person can attend all meetings and only one room is required).
